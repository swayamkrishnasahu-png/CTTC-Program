## MATPLOTLIB:USED FOR DATA VISULAIZATION

```python
import matplotlib.pyplot as plt
```

```python
#line plot
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.grid()
plt.show()
```

```python
plt.plot(x,y,c='k')
plt.title('line plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid()
plt.show()
```

```python
plt.bar(x,y,color='b')
plt.grid()
plt.show()
```

```python
plt.bar(x,y,color=['b','g','cyan','m','r'],width=2)
plt.grid()
plt.annotate('happy',(0.6,2.1),color='b')
plt.show()
```

```python
x=[1,3,5,7,9]
y=[2,4,6,8,10]
x1=[1,3,5,7,9]
y1=[2,4,6,8,10]
plt.bar(x,y)
plt.bar(x1,y1,width=0.2)
plt.show()
```

```python
#scatter plot
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x,y,s=700)#used to change the size
plt.show()
```

```python
plt.scatter(x,y,s=700,marker="^")#marker used for changing marker
plt.show()
```

```python
#pie chart
activities=['sleeping','playing','food','study','phone']
slices=[7,3,1,2,9]
plt.pie(slices,labels=activities)
plt.show()
```

```python
plt.pie(slices,labels=activities,colors=['r','g','b','cyan',
                                         'm'],autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(slices,labels=activities,colors=['r','g','b','cyan',
                                         'm'],autopct='%1.1f%%',explode=(0,0,1,0,0))
plt.show()
```

## SEABORN:STATISTICAL DATA VISUALIZATION

Continuous & Categorical: distplot, histplot, displot, kdeplot, boxplot, boxenplot, violinplot, barplot, swarmplot, stripplot, point plot

Continuous & Continuous: scatter plot, joint plot, heat map

Categorical (Only): count plot

All Columns: Pairplot

```python
#Nature of Data
#------->continuous data(ex-height,weight,income,income,temp etc)
#------->categorical data(ex-gender,binary values,boolean values(1,0),boolean(True,False))
import seaborn as sns
import pandas as pd
pd.__version__
sns.__version__
```

```python
auto=pd.read_csv(r"C:\Users\ASUS\Desktop\auto-mpg.csv")
auto
```

```python
#mpg-continuous,weight-continuous[scatter plot]
sns.scatterplot(x=auto.weight,y=auto.mpg)
plt.show()
```

```python
#reg plot(regression plot)
sns.regplot(x=auto.weight,y=auto.mpg)
plt.show()
```
