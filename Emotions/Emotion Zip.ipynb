```python
# importing packages: 
import cv2 
print (cv2.__version__)
import numpy as np 
import os          #helps to interact with the operating system
import matplotlib.pyplot as plt 
import pickle      # help to zip/ compress the image file in byte(binary) format. 
```

```python
# Accessing the images present inside "Emotion": 
# getcwd:- fetching the current working directory
img_dir= os.path.join(os.getcwd(),"Emotion")   
```

```python
# to checking the directory where currently we are working: 
print(os.getcwd())
```

```python
# creating a path to store the pickle files inside data folder. 
img_final= os.path.join(os.getcwd(),'data')
```

```python
# to store all the input(images) and output(labels) in different []. 
images= [] 
labels= [] 
```

```python
# performing image preprocessing and saving the files. 
# listdir: to list all folders/files present in any directory.
for i in os.listdir(img_dir): 
    for j in os.listdir(os.path.join(img_dir,i)):
        img_path= os.path.join(os.path.join(img_dir,i),j)
        img= cv2.imread(img_path) 
        img= cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) 
        img= cv2.resize(img,(48,48)) 
        images.append(img) 
        labels.append(i.split(' ')[0]) 
    print(i,'---> Completed')
```

```python
set(labels)
```

```python
images= np.array(images) 
labels= np.array(labels) 
```

```python
# Pickling the images and labels files. 
# wb = write binary , pickle file is always store in binary format.
#images.p= '.p' is the extension of pickle files.
#dump means to store 
with open(os.path.join(img_final,'images.p'),'wb') as f:
    pickle.dump(images,f) 

with open(os.path.join(img_final,'labels.p'),'wb') as f: 
    pickle.dump(labels,f) 
```
