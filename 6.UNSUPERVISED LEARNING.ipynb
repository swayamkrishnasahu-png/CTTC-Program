```python
#drop output column for unsupervised learning
iris=iris.drop('Class',axis=1)
iris
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#K-means clustering
from sklearn.cluster import KMeans
km=KMeans(n_clusters=3)
km.fit(iris)
```

```python
km_p=km.predict(iris)
km_p
```

```python
centroid=km.cluster_centers_
centroid
# sl sw pl pw #sentosa
              #versicolan
              #verginica
#the point on which they maintain equal dist
```

```python
sns.scatterplot(x=iris.sl,y=iris.pl,c=km_p)
plt.show()
```

## DBSCAN(Density Based Spatial Clustering With Noise)

```python
main=pd.read_csv(r"C:\Users\ASUS\Downloads\maintenance_data.csv")
main
```

```python
ip=main[['lifetime','pressureInd','moistureInd','temperatureInd']]
ip
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
ip=sc.fit_transform(ip)
ip
```

```python
from sklearn.cluster import DBSCAN
db=DBSCAN(eps=0.8,min_samples=5)
pred=db.fit_predict(ip)#fit_predict used to fit and predict data both
pred
```

```python
main['Prediction']=pred
main.head()
```
