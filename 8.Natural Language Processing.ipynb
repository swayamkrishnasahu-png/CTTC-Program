## TEXT PROCESSING STEPS-
1. TOKENIZATION-DIVIDING /SPLITTING OF A CORPUS INTO A DOCUMENT OR DOCUMENTS INTO WORDS

```python
doc="We are learning NLP. NLP is an an=mazing concept of AIML."
doc
```

```python
nltk.download('punkt_tab')
```

2. stemming-is a process of removing the suffices of a word into its root form by using nearest word
ex-playing-play

```python
from nltk import PorterStemmer
ps=PorterStemmer()
```

```python
ps.stem('learning')
```

```python
ps.stem('studied')
```

```python
from nltk import LancasterStemmer
ls=LancasterStemmer()
```

```python
ls.stem('Studied')
```

```python
ls.stem('went')
```

## 3.lemmatization:- is a process of reducing the suffices of a word to its root form using its meaning/dictinary word

```python
from nltk import WordNetLemmatizer
wl=WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
wl.lemmatize('went')
```

```python
wl.lemmatize('went','v')
```

```python
wl.lemmatize('running','v')
```

```python
wl.lemmatize('happier','a')
```

```python
wl.lemmatize('happier','v')
```

```python
wl.lemmatize('happier','n')
```

## 5.pos Tag(part of speech tags):-Labelling the words with their grammatical roles

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag
```

```python
sent="NLP is aamazing"
sent
```

```python
txt=word_tokenize(sent)
txt
```

```python
tags=nltk.pos_tag(txt)
tags
```

## 6.ne-chunk:means chunking or identifying the places ,persons or organizations

```python
text="Sundar Pichai is the CEO of Google Situated at New York"
```

```python
word=word_tokenize(text)
word
```

```python
txt_tag=nltk.pos_tag(word)
txt_tag
```

```python
nltk.download('maxent_ne_chunker_tab')
```

```python
nltk.download('words')
```

```python
from nltk import ne_chunk
chunk=ne_chunk(txt_tag)
chunk
```

## 7.Vectorizer:- It is a process of converting the words into raw numbers. 

1. count Vectorizer
2. tfidf Vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
```

```python
data=["I love singing and Painting"]
data
```

```python
x=cv.fit_transform(data)
x
```

```python
print(x)
```

```python
cv.vocabulary_
```

```python
print(x.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
```

```python
data=['I love Cricket','I love football','football is popular']
data
```

```python
x=tfidf.fit_transform(data)
print(x)
```

```python
tfidf.vocabulary_
```
