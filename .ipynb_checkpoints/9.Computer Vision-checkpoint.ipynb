## Color Combinations

```python
#black
r=np.full((2,2),0)
g=np.full((2,2),0)
b=np.full((2,2),0)
rgb=np.dstack((r,g,b))
plt.imshow(rgb)
plt.axis('off')
plt.show()
```

## cv2 attributes

```python
plt.imshow(img)
plt.show()
```

```python
#BGR2RGB
img1=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img1)
plt.axis('off')
plt.show()
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
img1=cv2.cvtColor(img,cv2.COLOR_BGR2RGBA)
plt.imshow(img1)
plt.axis('off')
plt.show()
img10=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img10)
plt.axis('off')
plt.show()
img100=cv2.cvtColor(img,cv2.COLOR_BGR2HLS)
plt.imshow(img100)
plt.axis('off')
plt.show()
#for gray
img1000=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.imshow(img1000,cmap='gray')
plt.axis('off')
plt.show()
```

```python
#blur image
blur=cv2.GaussianBlur(img1,(11,11),10)#3,3 or 5,5 or 9,9 it always remain odd cus if we give even value its centre will not be discoverable
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
#blur image
blur=cv2.GaussianBlur(img1,(1,201),100)#3,3 or 5,5 or 9,9 it always remain odd cus if we give even value its centre will not be discoverable
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
#face detection through ai
face=r"C:\Users\ASUS\Desktop\haarcascade_frontalface_default (1).xml"
model=cv2.CascadeClassifier(face)
img_path = r"C:\Users\ASUS\Pictures\images (2).jpg"
img = cv2.imread(img_path)
facepoints=model.detectMultiScale(img)
for x,y,w,h in facepoints:
    cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),10)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
#face detection through webcam
face=r"C:\Users\ASUS\Desktop\haarcascade_frontalface_default (1).xml"
cap=cv2.VideoCapture(0)

ret,frame=cap.read()
while ret:
    ret,frame=cap.read()
    
    if not ret:
        break
    model=cv2.CascadeClassifier(face)
    facepoints=model.detectMultiScale(frame)
    for x,y,w,h in facepoints:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(200,150,60),7)
    cv2.imshow('Face',frame)
    if cv2.waitKey(10)==ord('q'):
        break
cv2.destroyAllWindows()
cap.release()
```
